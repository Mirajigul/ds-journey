```python
def mean(xs: list[float]) -> float:
    """Return the arithmetic mean of xs."""
    return sum(xs) / len(xs)
```

```python
mean([1,2,3])
```

```python
def variance(xs: list[float]) -> float:
    """Return the sample variance of xs (divide by n - 1)."""
    m = mean(xs)          
    total = 0             
    if len(xs)<2 :
        raise ValueError("variance requires at least 2 values")
    for x in xs:          
        total = total +(x-m)**2    
    return total/(len(xs)-1)          
```

```python
variance([1,2,3])
```

```python
def std(xs: list[float]) -> float:
    """Return the sample standard deviation of xs."""
    return (variance(xs))**0.5
```

```python
data = [2, 4, 4, 4, 5, 5, 7, 9]
print(mean(data), variance(data), std(data))

import numpy as np
print(np.mean(data), np.var(data, ddof=1), np.std(data, ddof=1))

try:
    std([5])
except ValueError as e:
    print("Caught:", e)
```

```python

def analyze_products(products):
    for product in products:
        product["revenue"] = product["price"] * product["quantity"]

    total_revenue = sum(product["revenue"] for product in products)
    max_revenue = max(product["revenue"] for product in products)

    top_product = [
        value["name"] for value in products
        if value["revenue"] == max_revenue
    ][0]

    high_revenue_products = [
        value["name"] for value in products
        if value["revenue"] > 200
    ]

    return {
        "total_revenue": total_revenue,
        "top_product": top_product,
        "high_revenue_products": high_revenue_products
    }
```

```python
analyze_products(products)
```
