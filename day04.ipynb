```python
%%sql
CREATE TABLE Orders (
    order_id INTEGER,
    product_id INTEGER,
    quantity INTEGER,
    region TEXT
);
```

```python
%%sql
CREATE TABLE Products (
    product_id INTEGER,
    product_name TEXT,
    unit_price REAL
);
```

```python
%%sql
select * from Orders where region="North";
```

```python
%%sql
select order_id, quantity,region from Orders order by quantity DESC;
```

```python
%%sql
select order_id, quantity from Orders where region="South" order by quantity ASC;
```

```python
%%sql
select order_id, quantity from Orders order by quantity DESC limit 3;
```

```python
%%sql
select count(*), sum(quantity),avg(quantity) from Orders ; 
```

```python
%%sql
select count(*) as total_orders, 
    sum(quantity) as total_quantity,
    avg(quantity) as average_quantity
from Orders ; 
```

```python
%%sql
select region, sum(quantity)
from Orders
group by region;
```

```python
%%sql
select product_name, sum(quantity*unit_price) as total_revenue from Orders, Products
 where Orders.product_id=Products.product_id
group by product_name;
```

```python
%%sql
select Products.product_name, sum(Orders.quantity*Products.unit_price) as total_revenue from Orders
join Products
on Orders.product_id=Products.product_id
group by Products.product_name;
```

```python
%%sql
select Orders.region, sum(Orders.quantity*Products.unit_price) as total_revenue from Orders
join Products
on Orders.product_id=Products.product_id
group by Orders.region
order by total_revenue DESC;
```

```python
%%sql
select region,sum(quantity) as total_quantity 
from Orders
group by region
having total_quantity>5
order by total_quantity DESC;
```

```python
%%sql
select Orders.region, Products.product_name, sum(Orders.quantity*Products.unit_price) as total_revenue from Orders
join Products
on Orders.product_id=Products.product_id
group by Orders.region,Products.product_name
order by total_revenue DESC;
```
