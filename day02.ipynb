```python
import pandas as pd
import numpy as np

data = {
    "OrderID": [101, 102, 103, 104, 105, 105, 106, 107],
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor",
                "Mouse", "Mouse", "Desk", "Laptop"],
    "Category": ["Electronics", "Electronics", "Electronics",
                 "Electronics", "Electronics", "Electronics",
                 "Furniture", "Electronics"],
    "Quantity": [2, 5, 3, np.nan, 4, 4, 1, 1],
    "UnitPrice": [800, 20, 50, 250, 20, 20, 300, 800],
    "Region": ["North", "South", "North", "West",
               "South", "South", "East", "North"]
}

df = pd.DataFrame(data)
print(df)
```

```python
orders = pd.DataFrame({
    "OrderID": [101, 102, 103, 104, 105],
    "ProductID": [1, 2, 1, 3, 2],
    "Quantity": [2, 5, 1, 3, 4]
})

products = pd.DataFrame({
    "ProductID": [1, 2, 3, 4],
    "ProductName": ["Laptop", "Mouse", "Keyboard", "Monitor"],
    "UnitPrice": [800, 20, 50, 250]
})
```

```python
print(orders)
print(products)
```

```python
merged=pd.merge(orders,products,on="ProductID",how="inner")
print(merged)
```

```python
merge_left=pd.merge(orders,products,on="ProductID",how="left")
print(merge_left)
```

```python
order_test=pd.concat([
    orders, pd.DataFrame({
        "OrderID":[106], "ProductID":[5], "Quantity":[2]
    })
],ignore_index=True)
order_test
```

```python
merge_test_left=pd.merge(order_test,products,on="ProductID",how="left")
merge_test_left
```

```python
merge_right=pd.merge(order_test,products,on="ProductID",how="right")
merge_right
```

```python
merge_outer=pd.merge(order_test,products,on="ProductID",how="outer")
merge_outer
```

```python
merge_final=pd.merge(order_test, products,on="ProductID",how="inner")
merge_final["Revenue"]=merge_final["Quantity"]*merge_final["UnitPrice"].copy()
```
